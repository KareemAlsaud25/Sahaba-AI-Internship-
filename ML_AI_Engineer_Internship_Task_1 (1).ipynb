```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/amazon.csv")
```

```python
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

```python
features_to_drop = ['product_id', 'user_id', 'user_name', 'review_id', 'img_link', 'product_link','product_name', 'about_product', 'review_title', 'review_content']
df.drop(labels= features_to_drop, axis = 1, inplace = True)
```

```python
df.info()
```

```python
df.head()
```

```python
df['discounted_price'] = df['discounted_price'].str.replace(r'[₹,]', '', regex = True).astype('float')
df['actual_price'] = df['actual_price'].str.replace(r'[₹,]', '', regex = True).astype('float')
df['discount_percentage'] = df['discount_percentage'].str.replace(r'%', '', regex = True).astype('float')
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')
df['rating_count'] = pd.to_numeric(df['rating_count'].str.replace(',', ''), errors='coerce')
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df[['rating', 'rating_count']].describe()
```

```python
df['rating'] = df['rating'].fillna(df['rating'].mean())
df['rating_count'] = df['rating_count'].fillna(df['rating_count'].median())
```

```python
df.isnull().sum()
```

```python
df['category'] = df['category'].str.split('|').str[0]
```

```python
df['category'].unique()
```

```python
encoded_df = df.copy()
encoded_df = pd.get_dummies(encoded_df, columns = ['category'], drop_first = True, dtype = float)
```

```python
encoded_df.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
numerical_features = ['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']

encoded_df[numerical_features] = scaler.fit_transform(encoded_df[numerical_features])
```

```python
encoded_df.head()
```

```python
plt.figure(figsize=(8, 5))
# Plotting the raw rating values before scaling from your original df
plt.hist(df['rating'], bins=20, edgecolor='black', alpha=0.7)
plt.title('Distribution of Product Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
rating_skewness = df['rating'].skew()
print(f"The target column 'rating' Skewness: {rating_skewness}")
```

The data is highly imbalanced for a regression problem, as a skewness rate less than -1 means that the data is negatively skewwed to the left. This happened beauase most of the ratings are high.

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['actual_price'], df['rating'], alpha=0.6, color='royalblue')

plt.title('Product Price vs. Customer Rating')
plt.xlabel('Actual Price (₹)')
plt.ylabel('Rating (1-5)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['discount_percentage'], df['rating_count'], alpha=0.6, color='darkorange')

plt.title('Discount Percentage vs. Number of Ratings')
plt.xlabel('Discount Percentage (as Decimal)')
plt.ylabel('Rating Count (Popularity)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
category_ratings = df.groupby('category')['rating'].mean().sort_values()

plt.figure(figsize=(10, 6))
plt.barh(category_ratings.index, category_ratings.values, color='teal')

plt.title('Average Customer Rating by Category')
plt.xlabel('Average Rating (1-5)')
plt.ylabel('Product Category')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.show()
```
